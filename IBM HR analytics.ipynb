```python
df.drop_duplicates(inplace=True)
```

```python
df["Attrition_Flag"] = df["Attrition"].map({
    "Yes":1,
    "No":0
})
```

```python
df["OverTime_Flag"] = df["OverTime"].map({
    "Yes":1,
    "No":0
})
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 100],
    labels=["18-25", "26-35", "36-45", "46+"]
)
```

```python
df["TenureBand"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1,2,6,100],
    labels=[
        "New (0-2 Years)",
        "Mid (3-6 Years)",
        "Senior (7+ Years)"
    ]
)
```

```python
df["IncomeBand"] = pd.qcut(
    df["MonthlyIncome"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"])
```

```python
satisfaction_columns = [
    "EnvironmentSatisfaction",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance"
]

df["SatisfactionScore"] = (
    df[satisfaction_columns]
    .mean(axis=1)
    .round(2)
)
```

```python
df["PromotionStagnant"] = np.where(
    df["YearsSinceLastPromotion"] > 3,
    1,
    0
)
```

```python
df["ManagerStabilityRatio"] = np.where(
    df["YearsAtCompany"] > 0,
    (
        df["YearsWithCurrManager"] /
        df["YearsAtCompany"]
    ).round(2),
    0
)
```

```python
df["PromotionDelayRatio"] = np.where(
    df["YearsAtCompany"] > 0,
    (
        df["YearsSinceLastPromotion"] /
        df["YearsAtCompany"]
    ).round(2),
    0
)
```

```python
df["IncomePerJobLevel"] = (
    df["MonthlyIncome"] /
    df["JobLevel"]
).round(2)
```

```python
df["ExperienceBand"] = pd.cut(
    df["TotalWorkingYears"],
    bins=[-1,5,10,20,50],
    labels=[
        "0-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]
)
```

```python
df["LongDistanceEmployee"] = np.where(
    df["DistanceFromHome"] >= 20,
    1,
    0
)
```

```python
df["FrequentTraveller"] = np.where(
    df["BusinessTravel"] == "Travel_Frequently",
    1,
    0
)
```

```python
df["HighSalaryHike"] = np.where(
    df["PercentSalaryHike"] >= 20,
    1,
    0
)
```
